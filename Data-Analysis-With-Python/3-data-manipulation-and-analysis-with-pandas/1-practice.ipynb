```python
### Filling missing values with the mean of the column
import pandas as pd

dataframe2 = pd.read_csv('missing_data.csv')

dataframe2
```

```python
dataframe2.isnull()
```

```python
dataframe2.isnull().any(axis=0)
```

```python
dataframe2['avg_Value'] = dataframe2['Salary'].fillna(dataframe2['Salary'].mean())

```

```python
dataframe2.isnull().any(axis=0)
```

```python
dataframe2
```

```python
### Reraming Columns

dataframe2 = dataframe2.rename(columns={'PerformanceScore':'overAll_Score'})

dataframe2.head()
```

```python
### change the datatypes

dataframe2.dtypes
dataframe2.isnull().any(axis=0)
```

```python
### changing the datatype

dataframe2['Age_int'] = dataframe2['Age'].fillna(dataframe2['Age'].mean()).astype(int)

dataframe2
```

```python
### changing the values using lambda functuon
### lets say salary has been increased

dataframe2['New_Salary'] = dataframe2['Salary'].apply(lambda x : x*2)
```
