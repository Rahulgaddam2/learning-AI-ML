```python
import pandas as np

dataframe = np.read_csv('missing_data.csv')

dataframe
```

```python
### Data Grouping = Grouping means splitting data into categories based on a column.

### Data Aggregation = Aggregation means applying a summary function to each group.

```

```python
dataframe.head()
```

```python
### now based on salary column i want to do affregation and grouping

grouped_mean = dataframe.groupby('Salary')['ExperienceYears'].mean()

print(grouped_mean)
```

```python
### we can also group by more than 1 columns using a list

list_groupby = dataframe.groupby(['Department','ExperienceYears'])['Salary'].mean()

print(list_groupby)
```

```python
### Now aggregating multiple functions

aggre = dataframe.groupby('Department')['Salary'].agg(['mean','sum','count'])

print(aggre)
```
