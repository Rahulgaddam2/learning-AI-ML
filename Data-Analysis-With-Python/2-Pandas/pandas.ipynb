```python
import pandas as pd

### lets study pandas series
### A pandas series i a one-dimensional array like object that can hold any data type

data = [1,2,3,4,5,6]

series = pd.Series(data)
print(series) ### default indexs are 0,1,2,3,4,5
```

```python
### series from dictionary elemets
import pandas as pd
data = {
    'a' : 1,
    'b' : 2,
    'c' : 3
}

series_dic = pd.Series(data)

print(series_dic) ### Here indexs are key and the actual values are keyvales

### index = keys , values = series values
```

```python
import pandas as pd

### Here creating my own indexs

data = [10,20,30,40,50]

indexs = ['a','b','c','d','e'] ### Indexs should be same as values

pd.Series(data,index=indexs)

```

```python
### Now lets study data frames in pandas
### -> its two dimensional, mutable and potential

### create a dataframe from dic from list

### When you create a Pandas DataFrame from a dictionary:
### Dictionary keys → column names
### Dictionary values → column data 

import pandas as pd

data = {
    'name' : ['Rahul',"sushma","adhvay"],
    'age'  : [25,33,2],
    'city' : ['new-york','seattle','lynnwood']
}

dataFrame = pd.DataFrame(data)

print(dataFrame)
print(type(dataFrame))
```

```python
### create a dataframe form list of dic

import pandas as pd

data = [
    {'name' : 'Rahul', 'age' : 25 , 'city' : 'new-york'},
    {'name' : 'Sushma', 'age' : 33 , 'city' : 'seattle'},
    {'name' : 'Advay', 'age' : 2 , 'city' : 'lynnwood'}
]

pd.DataFrame(data)
```

```python
### Reading from a csv file

import pandas as pd

dataframe = pd.read_csv('employees.csv')

dataframe.head(5) ### gives the first 5 rows

dataframe.tail(5) ### gives the last 5 rows
```

```python
### Now most important accessing data from this dataframe

### pick all the elements which respective to name

dataframe['Name']
print(type(dataframe['Name'])) ## this will be series as it is an one-dimension array

```

```python
### using loc operation

dataframe.loc[1] ### accessing the row with index label 1
```

```python
dataframe.iloc[1] ### cares about physical position
```

```python
### accessing a specific element
### use at

dataframe.at[1,"Name"]
```

```python
### accessing a specific element with index
### use iat

dataframe.iat[2,2]
```

```python
### Data manipulation with data frames
### now adding a new column

dataframe['is_alive'] = ['yes','yes','yes','yes','yes','yes','yes','yes','yes','yes']
```

```python
### add age + 1 to everyone

dataframe['Age'] = dataframe['Age']+1
```
