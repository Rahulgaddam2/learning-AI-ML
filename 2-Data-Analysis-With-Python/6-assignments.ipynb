```python
### 1. Create a NumPy array of shape (5, 5) filled with random integers between 
# 1 and 20. Replace all the elements in the third column with 1.

import numpy as np

array = np.random.randint(2,20,(5,5))

print(array)
print(array.shape)
```

```python
### 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. 
# Replace the diagonal elements with 0.

array2 = np.arange(1,17).reshape(4,4)
print(array2) 

np.fill_diagonal(array2,0)

print(array2)
```

```python
### Create a NumPy array of shape (6, 6) with values from 1 to 36. 
# Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.

array3 = np.arange(1,37).reshape(6,6)
print(array3)

print(array3[1:6,2:5])
```
