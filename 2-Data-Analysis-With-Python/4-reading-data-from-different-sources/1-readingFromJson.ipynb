```python
### If JSON is in a Python variable (string)
import pandas as pd
from io import StringIO

data = '''[
    {"name": "Rahul", "age": 25, "salary": 70000},
    {"name": "Sushma", "age": 33, "salary": 180000}
]'''

### from io import StringIO
### StringIO converts a string into a file-like object in memory.
### It creates a fake file in RAM.
### So if we dont have real json file, but if we have a json data in python we can use this stringIo to convert data into file
### But the data must be string not list or anyother

data_file = StringIO(data)

dataframe = pd.read_json(data_file)

print(dataframe)
```

```python
### know to convert the jsonfile back to json string

dataframe.to_json()
```
