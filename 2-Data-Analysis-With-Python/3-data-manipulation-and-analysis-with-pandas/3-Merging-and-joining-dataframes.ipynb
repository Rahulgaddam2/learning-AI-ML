```python
### Merging -> Merging in Pandas means combining two DataFrames based on a common column

import pandas as pd

data1 = {
    'Employee-ID' : [1,2],
    'Salary' : [70000, 80000]
}

data2 = {
    'Employee-ID' : [1,2],
    'Name' : ["Rahul","Sushma"]
}

print(data1)
print(data2)

dataframe_data1 = pd.DataFrame(data1)
dataframe_data2 = pd.DataFrame(data2)

dataframe_data1
dataframe_data2
```

```python
### Now lets merge the data based on common column

merge = pd.merge(dataframe_data1,dataframe_data2, on="Employee-ID")

print(merge)
```

```python
### The 4 Join Types in Pandas

## Inner Join
## Left Join
## right Join
## Outer Join
```

```python
## lets take 2 data sets

data3 = {
    'Employee-ID' : [1,2,3,5],
    'Name' : ["Rahul","Sushma","Adhvay","Anjavva"]
}

data4 = {
    'Employee-ID' : [1,2,3,4],
    'Salary' : [100000,200000,300000,400000]
}

## inner join

dataframe_data3 = pd.DataFrame(data3)
dataframe_data4 = pd.DataFrame(data4)

merge = pd.merge(dataframe_data3,dataframe_data4, on='Employee-ID', how='inner')
### joins only same values in both Employee-ID

print(merge)

merge_outer = pd.merge(dataframe_data3,dataframe_data4, on='Employee-ID', how='outer')
### joins all the values
print(merge_outer)

merge_left = pd.merge(dataframe_data3,dataframe_data4,on='Employee-ID',how='left')
### Joins on the values from left
print(merge_left)

merge_right = pd.merge(dataframe_data3,dataframe_data4,on='Employee-ID',how='right')
### Joins on the values from right
print(merge_right)



```
