```python
import pandas as pd

dataframe = pd.read_csv('sales_data.csv')

dataframe
```

```python
total_product = dataframe.groupby('Product')['Total_Sales'].sum()

print(total_product)
```

```python
### Now lets create a bar graph

total_product.plot(kind='bar',color='blue')
```
