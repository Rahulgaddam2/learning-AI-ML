```python
### pyplot is the part of Matplotlib that helps us draw graphs using simple commands.

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [1,2,3,4,5]

### create a line plot

plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("Basic Line Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [6,7,8,9,10]

### create a customized line plot

plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=1,markersize=9)
plt.grid(True)
```

```python
### Now lets build multiplots

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.figure(figsize=(9,5))  ### graph size

plt.subplot(1,2,1) ### You are dividing the screen into 2 graphs side by side.
### 1 row , 2 cols , 1st position
plt.plot(x,y1,color='green')
plt.title('First plot')

plt.subplot(1,2,2)
plt.plot(x,y2,color = 'red')
plt.title('Second plot')
```

```python
### Bar plot
### A bar plot (bar chart) is a graph that uses rectangular bars to compare values of different categories.

categories = ['A','B','C','D']
values = [5,7,8,6]

### lets create a bar plot

plt.bar(categories,values,color='green')
```

```python
### Histogram

### A histogram is a graph used to show the distribution (spread) of numerical data.

### lets make a data

data = [1,1,2,3,3,3,3,4,4,4,5,5,5,5,5]  

### create a histogram

plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
### create a scatter plot

x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color='black',marker='X')
```

```python
### pipe chart

labels = ['A','B','C','D']
sizes = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode=(0.2,0,0,0)

### create a pipe chart

plt.pie(sizes,labels=labels,explode=explode)
```
