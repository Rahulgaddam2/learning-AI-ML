```python
### Working with sales data

import sqlite3
```

```python
Connection = sqlite3.connect('sale.db')
Connection
```

```python
cursor = Connection.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS sale (
               id INTEGER PRIMARY KEY,
               date TEXT NOT NULL,
               product TEXT NOT NULL,
               sales INTEGER,
               region TEXT
    )
''')

Connection.commit()

```

```python
### Insert data into the table
### Using python list and tuples

insertDate = [
    ('3/9/2026','Product1', 100, 'North'),
    ('3/8/2026','Product1', 500, 'North'),
    ('3/7/2026','Product1', 1000, 'North'),
    ('3/6/2026','Product1', 200, 'North'),
    ('3/5/2026','Product1', 50, 'North'),
]

Connection.executemany('''
    Insert Into sale(date,product,sales,region)
                       values(?,?,?,?)
''',insertDate)

Connection.commit()
```

```python
cursor.execute('Select * from sale')

rows = cursor.fetchall()

for row in rows:
    print(row)
```
