```python
### SQL is a programming language used to communicate with databases.
### SQL = the language used to talk to databases.

### SQLite is a lightweight database engine that uses SQL.
### SQLite is just one database system that uses SQL.
### SQLite is very widely used in embedded systems.

```

```python
### to use sqlite3 we have to import it

import sqlite3
```

```python
## connect to an sqlite database

Connection = sqlite3.connect('example.db')
Connection  ## the db file is created in the folder
```

```python
### creating a cursor

cursor = Connection.cursor()

### cursor is an object that lets your Python program execute SQL commands and interact with the database.
## Think of it as a control pointer or interface between Python and the database.
```

```python
### lets create a table

cursor = Connection.cursor()
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

### Commit the changes

Connection.commit()

### Without commit():
### The table creation might stay only in memory
### If the program closes, changes may disappear.
```

```python
### lets insert the data in sqlite table


cursor.execute('''
    Insert Into employees(name,age,department)
               values('rahul',25,'AI')
''')

cursor.execute('''
    Insert Into employees(name,age,department)
               values('Sushma',33,'AI')
''')

cursor.execute('''
    Insert Into employees(name,age,department)
               values('Adhvay',2,'AI')
''')

Connection.commit()
```

```python
### Query the data from the table

cursor.execute('''
    Select * From employees 
''')

rows = cursor.fetchall()

### print the query data

for row in rows:
    print(row)
```

```python
### Update the data in the table

cursor.execute('''
    UPDATE employees
    Set age=32
    Where name='Sushma'
''')

Connection.commit()
```
